## ขั้นตอนที่ 1: การนำเข้าไลบรารี pandas และการสำรวจรายชื่อแผ่นงานในไฟล์ Excel

เริ่มต้นสมุดงานใหม่ด้วยการนำเข้าไลบรารี pandas ในชื่อย่อ pd จากนั้นเปิดอ่านโครงสร้างไฟล์สมุดงาน Excel (nvc_store_sales.xlsx) ด้วยฟังก์ชัน pd.ExcelFile() และตรวจสอบรายชื่อแผ่นงานย่อยทั้งหมดภายในไฟล์ด้วยแอตทริบิวต์ .sheet_names

```python
import pandas as pd
excel_book = pd.ExcelFile('nvc_store_sales.xlsx')
excel_book.sheet_names
```

- `import pandas as pd`: นำเข้าไลบรารี pandas สำหรับจัดการข้อมูลตารางและกำหนดชื่อย่อเป็น pd (ต้องรันทุกครั้งเมื่อสร้างสมุดงานใหม่)
- `excel_book = pd.ExcelFile('nvc_store_sales.xlsx')`: เปิดอ่านโครงสร้างไฟล์สมุดงาน Excel เพื่อตรวจสอบแผ่นงานภายในโดยยังไม่ดึงข้อมูลทั้งตาราง
- `excel_book.sheet_names`: แอตทริบิวต์สำหรับแสดงรายชื่อแผ่นงาน (Sheets) ทั้งหมดที่มีอยู่ในไฟล์ Excel ในรูปแบบรายการ (List)

## ขั้นตอนที่ 3: การสำรวจชนิดข้อมูล (.dtypes) จำนวนค่าว่าง (.isna().sum) และค่าสถิติพื้นฐาน (.describe)

ทบทวนขั้นตอนการสำรวจโครงสร้างตารางพื้นฐานตั้งแต่ต้น โดยตรวจสอบชนิดข้อมูลด้วย .dtypes ตรวจนับจำนวนข้อมูลสูญหายแยกตามรายคอลัมน์ด้วย .isna().sum() และคำนวณค่าสถิติเชิงพรรณนาด้วย .describe()

```python
print(branch_df.dtypes)
print('--- จำนวนค่าว่างในแต่ละคอลัมน์ ---')
print(branch_df.isna().sum())
branch_df.describe()
```

- `print(branch_df.dtypes)`: แสดงชนิดข้อมูลของทั้ง 6 คอลัมน์ในตาราง branch_df
- `branch_df.isna().sum()`: นับจำนวนแถวที่มีค่าว่าง (NaN) ในแต่ละคอลัมน์ของตาราง branch_df
- `branch_df.describe()`: คำนวณค่าสถิติพื้นฐานของคอลัมน์ตัวเลข (quantity) เพื่อตรวจหาค่าต่ำสุดที่ติดลบและค่าสูงสุดที่สูงผิดปกติ

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์

## ขั้นตอนที่ 4: เทคนิคที่ 1: การจัดมาตรฐานรหัสสินค้าด้วย .str.strip() และ .str.upper()

ตัดอักขระช่องว่างด้านหน้าและด้านหลังรหัสสินค้า แล้วแปลงเป็นตัวพิมพ์ใหญ่เพื่อให้รหัสเดียวกันมีรูปแบบตรงกัน ก่อนตรวจจำนวนด้วย .value_counts()

```python
branch_df['product_id'] = branch_df['product_id'].str.strip().str.upper()
branch_df['product_id'].value_counts()
```

- `branch_df['product_id'].str.strip()`: ตัดอักขระช่องว่างส่วนเกินด้านหน้าและด้านหลังรหัสสินค้า เช่น ' p001 ' ให้เหลือ 'p001'
- `.str.upper()`: แปลงตัวอักษรภาษาอังกฤษทั้งหมดให้เป็นตัวพิมพ์ใหญ่มาตรฐานเดียวกัน เช่น 'p001' เป็น 'P001'
- `branch_df['product_id'].value_counts()`: ตรวจสอบผลการนับความถี่ของรหัสสินค้าภายหลังการจัดรูปแบบมาตรฐาน

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์


### ผลลัพธ์และคำอธิบาย

```text
product_id
P001    4
P002    4
P004    4
P006    4
P003    3
P005    3
```

รหัสสินค้าหลังจัดรูปแบบเป็นมาตรฐาน P001 ถึง P006 ครบ 6 รหัส โดยรหัสเดียวกันไม่ถูกนับแยกเพราะช่องว่างหรือตัวพิมพ์เล็ก–ใหญ่

**ข้อควรระวังในการเขียนคำสั่ง:** สามารถเขียนคำสั่งต่อเนื่องกัน (Method Chaining) เป็น .str.strip().str.upper() ในบรรทัดเดียวได้ แต่ต้องอย่าลืมกำหนดค่ากลับเข้าสู่คอลัมน์เดิมด้วยเครื่องหมาย =

**คำถามตรวจสอบความเข้าใจ:** หากรหัส p001 และ P001 เป็นสินค้าเดียวกัน แต่ไม่จัดรูปแบบก่อนนับจำนวน จะเกิดผลอย่างไร


## ขั้นตอนที่ 5: เทคนิคที่ 2: การจัดกลุ่มคำศัพท์ที่เขียนต่างกันให้เป็นมาตรฐานเดียวด้วย .replace()

สร้างพจนานุกรม (Dictionary) จับคู่คำศัพท์ที่เขียนแตกต่างกันให้เป็นคำมาตรฐานเดียวกัน แล้วใช้ฟังก์ชัน .replace() แทนที่ค่าในคอลัมน์ payment_method พร้อมตรวจสอบจำนวนรายการแต่ละช่องทาง

```python
payment_map = {'โอน': 'พร้อมเพย์', 'โอนเงิน': 'พร้อมเพย์', 'QR PromptPay': 'พร้อมเพย์', 'สด': 'เงินสด'}
branch_df['payment_method'] = branch_df['payment_method'].str.strip().replace(payment_map)
branch_df['payment_method'].value_counts()
```

- `payment_map = {'โอน': 'พร้อมเพย์', ...}`: กำหนดพจนานุกรมระบุคำเดิมที่ต้องการเปลี่ยน (Key) และคำมาตรฐานที่ต้องการใช้แทน (Value)
- `.str.strip().replace(payment_map)`: ตัดช่องว่างส่วนเกินและแทนที่คำศัพท์ในคอลัมน์ payment_method ตามพจนานุกรมที่กำหนด
- `branch_df['payment_method'].value_counts()`: ตรวจนับจำนวนรายการแยกตามช่องทางการชำระเงินภายหลังการจัดมาตรฐานคำศัพท์

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์


### ผลลัพธ์และคำอธิบาย

```text
payment_method
พร้อมเพย์    14
เงินสด        8
```

ช่องทางการชำระเงินที่เคยแยกเป็น 6 รูปแบบ ถูกรวมกลุ่มเป็นมาตรฐานเดียวกันเหลือเพียง 2 ช่องทางหลัก ได้แก่ 'พร้อมเพย์' (14 รายการ) และ 'เงินสด' (8 รายการ)

**ข้อควรระวังในการเขียนคำสั่ง:** การสะกดคำต้นทางในพจนานุกรม (Key) ต้องตรงกับข้อความในตารางทุกตัวอักษร จึงควรใช้ .str.strip() ตัดช่องว่างก่อนเรียกใช้ .replace()

**คำถามตรวจสอบความเข้าใจ:** เทคนิคการใช้ .replace() ด้วยพจนานุกรมมีข้อดีกว่าการเขียนคำสั่งแก้ไขทีละคำอย่างไร


## ขั้นตอนที่ 6: เทคนิคที่ 3: การจัดการค่าว่างด้วยการเติมค่ามาตรฐาน (.fillna) แทนการลบแถว

เติมส่วนลดที่ว่างด้วยข้อความ 0 บาท และจำนวนที่ว่างด้วยค่าทดแทน 1 แก้วตามสมมติฐานที่กำหนดเฉพาะชุดฝึกนี้ เพื่อฝึกใช้ .fillna() และคงจำนวนแถวไว้ ต้องแยกค่าทดแทนออกจากค่าที่ตรวจสอบยืนยันแล้วในการใช้งานจริง

```python
branch_df['discount_text'] = branch_df['discount_text'].fillna('0 บาท')
branch_df['quantity'] = branch_df['quantity'].fillna(1)
branch_df.isna().sum()
```

- `branch_df['discount_text'].fillna('0 บาท')`: เติมข้อความ '0 บาท' ลงในแถวที่มีค่าว่างของคอลัมน์ส่วนลด แทนการลบแถวข้อมูลทิ้ง
- `branch_df['quantity'].fillna(1)`: เติมค่าเริ่มต้นเท่ากับ 1 แก้ว ลงในแถวที่ไม่ได้ระบุจำนวนชิ้น
- `branch_df.isna().sum()`: ตรวจสอบยืนยันว่าไม่มีค่าว่าง (NaN) หลงเหลืออยู่ในตารางข้อมูล

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์

**เงื่อนไขก่อนใช้คำสั่ง:** กติกาสมมติสำหรับชุดฝึกนี้เท่านั้น: กำหนดให้ค่าว่างส่วนลดหมายถึง 0 บาท และจำนวนที่ไม่ระบุใช้ค่าทดแทน 1 แก้วเพื่อฝึก .fillna() ค่าทดแทนไม่ใช่ยอดขายที่ยืนยันแล้ว ข้อมูลจริงต้องตรวจหลักฐานและบันทึกการเติมค่า


### ผลลัพธ์และคำอธิบาย

```text
order_id          0
sale_date         0
product_id        0
quantity          0
discount_text     0
payment_method    0
```

จำนวนค่าว่างในทุกคอลัมน์กลายเป็น 0 รายการ โดยที่ตารางยังคงมีข้อมูลครบทั้ง 22 แถว ไม่สูญเสียรายการขายไปจากการลบแถว

**ข้อควรระวังในการเขียนคำสั่ง:** ค่าว่างไม่ได้หมายถึงศูนย์หรือหนึ่งเสมอไป ต้องตรวจสอบความหมายกับแหล่งข้อมูลก่อนเติมค่า และบันทึกว่ารายการใดได้รับการเติมค่า

**คำถามตรวจสอบความเข้าใจ:** เมื่อใดที่เราควรใช้ .fillna() แทนการใช้ .dropna() ในการจัดเตรียมข้อมูล


## ขั้นตอนที่ 7: เทคนิคที่ 4: การตัดหน่วยข้อความออกจากตัวเลขและแปลงชนิดข้อมูล (.str.replace & .astype)

สกัดเฉพาะตัวเลขออกจากคอลัมน์ discount_text โดยใช้ .str.replace() ลบคำว่า ' บาท' ออก แล้วแปลงชนิดข้อมูลจากข้อความให้เป็นตัวเลขทศนิยม (float) ด้วย .astype(float) เก็บในคอลัมน์ใหม่ชื่อ discount

```python
branch_df['discount'] = branch_df['discount_text'].str.replace(' บาท', '', regex=False).astype(float)
branch_df[['order_id', 'discount_text', 'discount']].head(6)
```

- `.str.replace(' บาท', '', regex=False)`: ค้นหาคำว่า ' บาท' ในข้อความแล้วแทนที่ด้วยค่าว่าง (ตัดคำว่า ' บาท' ออก)
- `.astype(float)`: แปลงชนิดข้อมูลจากข้อความตัวเลข (เช่น '10') ให้เป็นข้อมูลตัวเลขทศนิยม (10.0) ที่คำนวณได้
- `branch_df[['order_id', 'discount_text', 'discount']].head(6)`: แสดงเปรียบเทียบคอลัมน์ข้อความเดิมกับคอลัมน์ตัวเลขใหม่ 6 แถวแรก

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์


### ผลลัพธ์และคำอธิบาย

| ลำดับแถว | order_id | discount_text | discount |
| --- | --- | --- | --- |
| 0 | BR-101 | 10 บาท | 10.0 |
| 1 | BR-102 | 0 บาท | 0.0 |
| 2 | BR-103 | 5 บาท | 5.0 |
| 3 | BR-104 | 0 บาท | 0.0 |
| 4 | BR-104 | 0 บาท | 0.0 |
| 5 | BR-105 | 5 บาท | 5.0 |

ได้คอลัมน์ใหม่ discount ที่มีชนิดข้อมูลเป็น float64 (เช่น 10.0, 0.0, 5.0) พร้อมสำหรับนำไปหักลบออกจากยอดขายรวมได้อย่างถูกต้อง

**ข้อควรระวังในการเขียนคำสั่ง:** ต้องตัดอักขระที่ไม่ใช่ตัวเลข (เช่น ' บาท' หรือเครื่องหมายจุลภาค ',') ออกให้หมดก่อนเรียกใช้ .astype(float) มิฉะนั้นจะเกิดข้อผิดพลาด ValueError

**คำถามตรวจสอบความเข้าใจ:** เหตุใดเราจึงต้องแปลงคอลัมน์ส่วนลดให้เป็นชนิด float ก่อนนำไปคำนวณยอดขายสุทธิ


## ขั้นตอนที่ 8: เทคนิคที่ 5: การแก้ไขค่าติดลบด้วยค่าสัมบูรณ์ (.abs) และการจำกัดเพดานค่า (.clip)

สาธิต .abs() เพื่อแปลงค่าติดลบเป็นค่าบวก และ .clip(upper=10) เพื่อจำกัดจำนวนไม่เกิน 10 แก้วตามกติกาสมมติของแบบฝึก จากนั้นแปลงเป็นจำนวนเต็มด้วย .astype(int) การจำกัดเพดานไม่ใช่การกู้คืนค่าจริงจากต้นทาง

```python
branch_df['quantity'] = branch_df['quantity'].abs().clip(upper=10).astype(int)
branch_df['quantity'].describe()
```

- `.abs()`: ฟังก์ชันค่าสัมบูรณ์ (Absolute Value) เปลี่ยนตัวเลขติดลบ (เช่น -2, -3) ให้เป็นตัวเลขบวก (2, 3)
- `.clip(upper=10)`: ฟังก์ชันจำกัดขอบเขตข้อมูล โดยปรับค่าที่สูงเกินเพดาน 10 ให้ลงมาอยู่ที่ค่าสูงสุดที่ยอมรับได้คือ 10
- `.astype(int)`: แปลงชนิดข้อมูลจำนวนชิ้นจากทศนิยม (float64) ให้เป็นจำนวนเต็ม (int64)
- `branch_df['quantity'].describe()`: ตรวจสอบค่าสถิติพื้นฐานของคอลัมน์จำนวนชิ้นภายหลังการปรับแก้

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์

**เงื่อนไขก่อนใช้คำสั่ง:** กติกาสมมติสำหรับฝึกคำสั่ง: ถือว่าค่าติดลบเกิดจากเครื่องหมายผิด และจำกัดจำนวนสูงสุดไว้ที่ 10 แก้วเพื่อสาธิต .clip() การเปลี่ยน 50 เป็น 10 เป็นการจำกัดเพดาน ไม่ใช่การยืนยันยอดขายจริง หากยังตรวจต้นทางไม่ได้ ให้แยกรายการไว้ตรวจสอบ

```text
count    22.000000
mean      2.500000
std       1.896111
min       1.000000
25%       2.000000
50%       2.000000
75%       3.000000
max      10.000000
```

คอลัมน์ quantity มีค่าต่ำสุด (min) เท่ากับ 1 แก้ว และค่าสูงสุด (max) เท่ากับ 10 แก้ว ค่าเฉลี่ยอยู่ที่ประมาณ 2.55 แก้วต่อรายการ โดยไม่มีค่าติดลบหรือค่าที่สูงเกินเกณฑ์หลงเหลืออยู่

**ข้อควรระวังในการเขียนคำสั่ง:** ห้ามเปลี่ยนค่าติดลบหรือค่าที่สูงผิดสังเกตโดยอัตโนมัติในข้อมูลจริง เพราะอาจเป็นรายการคืนสินค้าหรือยอดขายจริง ต้องตรวจสอบหลักฐานและบันทึกเหตุผลก่อนแก้ไข

**คำถามตรวจสอบความเข้าใจ:** หากจำนวน 50 แก้วยังไม่มีหลักฐานยืนยันว่าเป็นข้อมูลผิด ควรเปลี่ยนเป็น 10 ทันทีหรือไม่ เพราะเหตุใด และ .clip() ต่างจากการลบแถวอย่างไร


## ขั้นตอนที่ 9: เทคนิคที่ 6: การลบข้อมูลซ้ำโดยระบุคอลัมน์คีย์หลัก (.drop_duplicates(subset=...))

ลบแถวที่มีรหัสคำสั่งซื้อ (order_id) ซ้ำซ้อนกัน โดยระบุพารามิเตอร์ subset=['order_id'] เพื่อตรวจสอบการซ้ำเฉพาะคอลัมน์คีย์หลัก และเก็บแถวแรกไว้ด้วย keep='first' ลงในตารางใหม่ชื่อ branch_clean

```python
branch_clean = branch_df.drop_duplicates(subset=['order_id'], keep='first').copy()
print('ขนาดตารางก่อนลบซ้ำ:', branch_df.shape)
print('ขนาดตารางหลังลบซ้ำ:', branch_clean.shape)
```

- `branch_df.drop_duplicates(...)`: ฟังก์ชันสำหรับลบแถวข้อมูลที่มีการบันทึกซ้ำซ้อน
- `subset=['order_id']`: ระบุให้ตรวจสอบความซ้ำซ้อนเฉพาะคอลัมน์รหัสคำสั่งซื้อ (order_id) แม้คอลัมน์อื่นอาจบันทึกต่างกันเล็กน้อย
- `keep='first'`: เก็บรักษาแถวแรกของรหัสคำสั่งซื้อนั้นไว้ และลบแถวที่ซ้ำตามมาออก
- `.copy()`: สร้างสำเนาตารางแยกออกมาเป็นตัวแปร branch_clean อย่างอิสระ

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์

**เงื่อนไขก่อนใช้คำสั่ง:** กำหนดให้ 1 order_id แทน 1 รายการขายในชุดฝึกนี้ จึงเก็บแถวแรกเมื่อพบรหัสซ้ำ ข้อมูลจริงที่หนึ่งคำสั่งซื้อมีหลายสินค้า หรือแถวซ้ำมีค่าขัดแย้งกัน ต้องตรวจสอบก่อนลบ

ขนาดตารางลดลงจาก (22, 7) เหลือ (20, 7) เนื่องจากรหัสคำสั่งซื้อที่ซ้ำกัน 2 คู่ (BR-104 และ BR-112) ถูกลบแถวที่ซ้ำออกเหลือรหัสละ 1 แถว

**ข้อควรระวังในการเขียนคำสั่ง:** หากไม่ระบุ subset=['order_id'] คำสั่ง .drop_duplicates() จะลบเฉพาะแถวที่ซ้ำกันทุกคอลัมน์เท่านั้น ซึ่งอาจไม่ลบรหัสคำสั่งซื้อเดียวกันที่พนักงานบันทึกช่องทางชำระเงินต่างกัน

**คำถามตรวจสอบความเข้าใจ:** ตารางควรลดจาก 22 เหลือ 20 แถว หากหนึ่งคำสั่งซื้อมีหลายรายการสินค้า ยังใช้ order_id เพียงคอลัมน์เดียวเพื่อลบซ้ำได้หรือไม่


## ขั้นตอนที่ 10: เทคนิคที่ 7: การแปลงชนิดข้อมูลวันที่ด้วย pd.to_datetime()
