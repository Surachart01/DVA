## ขั้นที่ 13: จัดตารางหลักฐานการนำเสนอ

รวบรวมชื่อ ยอดเงิน และสัดส่วนในตารางหนึ่งแถว

```python
evidence = pd.DataFrame({'เมนู':[top_menu], 'ยอดขาย (บาท)':[top_revenue], 'สัดส่วนยอดขาย (%)':[share]})
```

- `pd.DataFrame`: สร้างตารางจากข้อมูลที่คำนวณแล้ว
- `[...]`: รายการค่าของแต่ละคอลัมน์

เขียนคำสั่งและกด Shift + Enter ก่อนอ่านผลด้านล่าง


````python
evidence = pd.DataFrame({'เมนู':[top_menu], 'ยอดขาย (บาท)':[top_revenue], 'สัดส่วนยอดขาย (%)':[share]})
````

### การตรวจผล

evidence เก็บข้อมูลสำหรับตรวจคำอธิบายและข้อเสนอแนะ โดยยังไม่แสดงตาราง

**เมื่อเกิดข้อผิดพลาด:** ตรวจชื่อคอลัมน์ ตัวแปร และวงเล็บ เรียกใช้เซลล์ก่อนหน้าตามลำดับก่อนดำเนินการต่อ

```python
evidence.to_csv('W13_หลักฐานนำเสนอ.csv', index=False, encoding='utf-8-sig')
```

- `to_csv(...)`: เขียนตารางเป็นไฟล์ CSV ตามชื่อไฟล์ที่กำหนด
- `index=False`: ไม่เพิ่มคอลัมน์ดัชนี
- `encoding`: รูปแบบข้อความรองรับภาษาไทย

เขียนคำสั่งและกด Shift + Enter ก่อนอ่านผลด้านล่าง


````python
evidence.to_csv('W13_หลักฐานนำเสนอ.csv', index=False, encoding='utf-8-sig')
````

### การตรวจผล

```text
ไม่มีผลลัพธ์แสดงใต้เซลล์ และไม่มีข้อความผิดพลาด
```

ดาวน์โหลดไฟล์จาก Files และใช้ตรวจข้อความบนสไลด์ ไม่ใช้ยอดขายชุดเล็กนี้รับรองอนาคต

**เมื่อเกิดข้อผิดพลาด:** ตรวจชื่อคอลัมน์ ตัวแปร และวงเล็บ เรียกใช้เซลล์ก่อนหน้าตามลำดับก่อนดำเนินการต่อ

**ตรวจความเข้าใจ:** อธิบายหน้าที่ของคำสั่ง และเปรียบเทียบผลลัพธ์กับตัวอย่าง
