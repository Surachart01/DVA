## วันจันทร์ — วันจันทร์: ทบทวน Pipeline และตรวจความพร้อม

- ทบทวนงานทั้ง 4 ขั้นจากตัวอย่างฝึกและโครงงาน
- ดูตัวอย่างวิธีอธิบายผลด้วยหลักฐาน
- ตรวจ Notebook และแฟ้มงานทีละรายการ
- เตรียมสอบปลายภาคในวันอังคาร


## วันจันทร์ — ทบทวนความเชื่อมโยงของ Pipeline

- Get Data: คำถาม แหล่งข้อมูล และความหมายคอลัมน์
- ETL: การตรวจปัญหา การจัดเตรียม และการตรวจผล
- Modeling: ความสัมพันธ์ สูตรคำนวณ และตัวชี้วัด
- Visualization: การเลือกกราฟ การสื่อสารผล และข้อจำกัด


## วันจันทร์ — ดูตัวอย่างการอธิบายผลทีละรายการ

- เริ่มจากคำถามและความหมายของข้อมูลที่ใช้
- ระบุคำสั่งหรือวิธีที่เลือก พร้อมเหตุผล
- ชี้ผลจากตารางหรือกราฟและอธิบายหน่วยให้ชัดเจน
- สรุปสิ่งที่ข้อมูลรองรับและข้อจำกัดของคำอธิบาย


## วันจันทร์ — ตรวจ Notebook และแฟ้มงานทีละรายการ

- ตรวจชื่อไฟล์และลำดับเซลล์ที่จำเป็นต่อการรัน
- ตรวจผลของการอ่านข้อมูล การจัดเตรียม และการคำนวณ
- ตรวจว่าตัวเลขใน Dashboard และสไลด์ตรงกับผลใน Notebook
- ตรวจหลักฐานส่งงานและรายการที่ยังขาดตามเอกสารเกณฑ์เดิม


## วันจันทร์ — สรุปความพร้อมก่อนสอบปลายภาค

- อธิบายขั้นตอนและผลจากงานเดิมด้วยตนเองได้
- ระบุหัวข้อที่ต้องทบทวนต่อและเตรียมอุปกรณ์ที่กำหนด
- ทราบเวลา รูปแบบ และช่องทางการสอบ
- การทบทวนและตรวจแฟ้มไม่คิดคะแนนซ้ำ


## วันอังคาร — วันอังคาร: คำชี้แจงสอบปลายภาค

- สอบรายบุคคลเป็นเวลา 120 นาที
- คะแนนรวม 20 คะแนน
- ปรนัย 40 ข้อ รวม 10 คะแนน
- อัตนัย 2 ข้อ รวม 10 คะแนน


## วันอังคาร — ดำเนินการสอบและตรวจความครบถ้วนก่อนส่ง

- อ่านคำชี้แจงของแต่ละส่วนและจัดสรรเวลาด้วยตนเอง
- ตอบข้อสอบรายบุคคลตามระเบียบการสอบ
- ตรวจชื่อผู้เรียน จำนวนคำตอบ และหลักฐานที่ต้องส่ง
- ส่งตามช่องทางและเวลาที่ผู้สอนกำหนด


## วันอังคาร — สะท้อนการเรียนรู้หลังสอบ

- ทักษะใดที่พัฒนาขึ้นและมีหลักฐานจากงานใด
- ข้อจำกัดหรือจุดตรวจใดที่ต้องระวังเมื่อใช้ข้อมูล
- จะนำทักษะใดไปใช้ในการฝึกงานหรือการทำงาน
- การสะท้อนการเรียนรู้ไม่คิดคะแนนปลายภาคซ้ำ
