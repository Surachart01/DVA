## ขั้นตอนที่ 2: การอ่านไฟล์ข้อมูล CSV และกำหนดชนิดข้อมูลหมายเลขโทรศัพท์ด้วยคำสั่ง pd.read_csv()

ใช้ฟังก์ชัน pd.read_csv() เพื่ออ่านข้อมูลจากไฟล์ CSV เข้าสู่ตัวแปรตาราง df โดยระบุพารามิเตอร์ dtype เพื่อกำหนดชนิดข้อมูลของคอลัมน์หมายเลขโทรศัพท์ให้เป็นข้อความ (string)

```python
df = pd.read_csv('nvc_tech_store.csv', dtype={'customer_phone': 'string'})
df
```

- `df =`: กำหนดตัวแปรชื่อ df สำหรับจัดเก็บตารางข้อมูลต้นฉบับ
- `pd.read_csv('nvc_tech_store.csv', ...)`: ฟังก์ชันสำหรับอ่านไฟล์ข้อมูล CSV ของร้านอุปกรณ์ไอที
- `dtype={'customer_phone': 'string'}`: พารามิเตอร์กำหนดชนิดข้อมูลของคอลัมน์ customer_phone ให้เป็นข้อความ (string) เพื่อรักษาเลข 0 นำหน้า

## ขั้นตอนที่ 5: การตรวจสอบขนาดของตารางข้อมูลด้วยคำสั่ง df.shape

ใช้คำสั่ง .shape เพื่อตรวจสอบขนาดของตารางข้อมูลในรูปแบบคู่ลำดับ (จำนวนแถว, จำนวนคอลัมน์)

```python
df.shape
```

- `df`: ตัวแปรตารางข้อมูลต้นฉบับ
- `.shape`: คำสั่งตรวจสอบจำนวนแถวและจำนวนคอลัมน์ของตารางโดยไม่ต้องใส่วงเล็บต่อท้าย

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์

## ขั้นตอนที่ 7: การตรวจสอบโครงสร้างตารางและชนิดข้อมูลด้วยคำสั่ง df.info()

ใช้คำสั่ง .info() เพื่อตรวจสอบโครงสร้างตารางและชนิดข้อมูล ประกอบด้วยจำนวนแถวทั้งหมด รายชื่อคอลัมน์ จำนวนข้อมูลที่ไม่เป็นค่าว่าง (Non-Null Count) และชนิดข้อมูล (Dtype)

```python
df.info()
```

- `df`: ตัวแปรตารางข้อมูลต้นฉบับ
- `.info()`: คำสั่งแสดงข้อมูลสรุปโครงสร้างตาราง ชนิดข้อมูล และจำนวนข้อมูลที่มีอยู่ในแต่ละคอลัมน์

ใช้คำสั่ง .describe() เพื่อคำนวณค่าสถิติพื้นฐานของคอลัมน์ตัวเลข ได้แก่ จำนวนข้อมูล (count) ค่าเฉลี่ย (mean) ส่วนเบี่ยงเบนมาตรฐาน (std) ค่าต่ำสุด (min) ค่าควอไทล์ และค่าสูงสุด (max)

ตรวจพบค่าผิดปกติ (Outlier) 2 รายการ ได้แก่ คอลัมน์ quantity มีค่าต่ำสุด (min) เท่ากับ -2.0 ชิ้น และคอลัมน์ total_price มีค่าสูงสุด (max) เท่ากับ 99,000.0 บาท ส่งผลให้ค่าเฉลี่ยคลาดเคลื่อนจากความเป็นจริง (3,613.82 บาท)

**ข้อควรระวังในการเขียนคำสั่ง:** ไม่ควรนำค่าเฉลี่ย (mean) ไปสรุปผลทันทีโดยไม่ตรวจสอบค่าต่ำสุด (min) และค่าสูงสุด (max) ก่อน เนื่องจากค่าผิดปกติจะส่งผลให้ค่าเฉลี่ยคลาดเคลื่อนจากความเป็นจริง

**คำถามตรวจสอบความเข้าใจ:** ค่าต่ำสุด (min) ของคอลัมน์ quantity และค่าสูงสุด (max) ของคอลัมน์ total_price มีความผิดปกติในเชิงธุรกิจอย่างไร


## ขั้นตอนที่ 9: การนับจำนวนค่าว่างแยกตามรายคอลัมน์ด้วยคำสั่ง df.isna().sum()

ใช้คำสั่ง .isna() ร่วมกับ .sum() เพื่อตรวจสอบและรวมจำนวนค่าว่าง (Missing Values) แยกตามแต่ละคอลัมน์

```python
df.isna().sum()
```

- `df.isna()`: คำสั่งตรวจสอบว่าข้อมูลแต่ละตำแหน่งเป็นค่าว่างหรือไม่ (ให้ผลลัพธ์เป็น True หรือ False)
- `.sum()`: คำสั่งรวมจำนวนตำแหน่งที่เป็นค่าว่างแยกตามรายคอลัมน์

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์

สรุปจำนวนค่าว่างรายคอลัมน์ได้แก่ customer_phone มีค่าว่าง 2 แถว, quantity มีค่าว่าง 1 แถว และ total_price มีค่าว่าง 2 แถว (รวมทั้งตารางมีค่าว่าง 5 จุด)

**ข้อควรระวังในการเขียนคำสั่ง:** หากใช้คำสั่ง df.isna() เพียงอย่างเดียวโดยไม่ต่อท้ายด้วย .sum() โปรแกรมจะแสดงตารางค่าความจริง (True/False) ทั้ง 36 แถว ซึ่งไม่สะดวกต่อการสรุปจำนวน

**คำถามตรวจสอบความเข้าใจ:** คอลัมน์ใดบ้างที่ปรากฏค่าว่าง และมีจำนวนค่าว่างคอลัมน์ละกี่แถว
