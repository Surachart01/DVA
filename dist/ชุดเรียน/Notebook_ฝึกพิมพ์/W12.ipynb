## ขั้นที่ 12: สร้างพื้นที่ Dashboard

สร้างพื้นที่สองแถว โดยแถวแรกมีตัวเลขสรุปและกราฟแท่ง แถวที่สองเป็นกราฟเส้น

```python
fig = make_subplots(rows=2, cols=2, specs=[[{'type':'indicator'}, {'type':'xy'}], [{'type':'xy', 'colspan':2}, None]])
```

- `rows=2 / cols=2`: พื้นที่สองแถวสองคอลัมน์
- `specs`: ชนิดของกราฟแต่ละตำแหน่ง
- `colspan=2`: กราฟเส้นใช้ความกว้างสองคอลัมน์
- `None`: ไม่สร้างกราฟใหม่ในช่องที่ถูกครอบ
- `type='indicator'`: พื้นที่แสดงตัวเลขสรุป
- `type='xy'`: พื้นที่กราฟที่มีแกน x และ y

เขียนคำสั่งและกด Shift + Enter ก่อนอ่านผลด้านล่าง


### การตรวจผล

```text
ไม่มีผลลัพธ์แสดงใต้เซลล์ และไม่มีข้อความผิดพลาด
```

พื้นที่ยังไม่มีข้อมูล กำหนดตัวแปร fig เพื่อเพิ่มข้อมูลในขั้นต่อไป

**เมื่อเกิดข้อผิดพลาด:** ตรวจชื่อคอลัมน์ ตัวแปร และวงเล็บ เรียกใช้เซลล์ก่อนหน้าตามลำดับก่อนดำเนินการต่อ

**ตรวจความเข้าใจ:** อธิบายหน้าที่ของคำสั่ง และเปรียบเทียบผลลัพธ์กับตัวอย่าง


## ขั้นที่ 13: เพิ่มตัวเลขสรุป

add_trace() เพิ่มชุดข้อมูลหนึ่งชุดลงพื้นที่กราฟ โดยขั้นนี้ใช้ชุดข้อมูลตัวเลขสรุป

```python
fig.add_trace(go.Indicator(mode='number', value=total, title={'text':'ยอดขายรวม (บาท)'}), row=1, col=1)
```

- `go.Indicator`: ชุดข้อมูลตัวเลขสรุป
- `value=total`: ใช้ยอดขายรวมที่คำนวณแล้ว
- `row=1, col=1`: ตำแหน่งตัวเลขสรุป
- `add_trace(...)`: เพิ่มชุดข้อมูลลงพื้นที่กราฟที่ระบุ
- `mode='number'`: แสดงเฉพาะตัวเลขสรุป
- `title={'text':...}`: ตั้งข้อความกำกับตัวเลขใน key text

เขียนคำสั่งและกด Shift + Enter ก่อนอ่านผลด้านล่าง


### การตรวจผล

ผลลัพธ์เป็นกราฟแบบโต้ตอบ ตรวจเทียบใน Notebook ตัวอย่าง หรือเปิดไฟล์ `ตัวอย่าง/W12_ขั้น13.html` ในชุดเรียนย้อนหลัง

ตัวเลขสรุปแสดง 4,885 บาท กราฟส่วนอื่นยังไม่มีข้อมูล

**เมื่อเกิดข้อผิดพลาด:** ตรวจชื่อคอลัมน์ ตัวแปร และวงเล็บ เรียกใช้เซลล์ก่อนหน้าตามลำดับก่อนดำเนินการต่อ

**ตรวจความเข้าใจ:** อธิบายหน้าที่ของคำสั่ง และเปรียบเทียบผลลัพธ์กับตัวอย่าง


## ขั้นที่ 14: เพิ่มกราฟแท่ง

เพิ่มกราฟแท่งแยกเมนูในช่องขวาบน

```python
fig.add_trace(go.Bar(x=by_menu.index, y=by_menu.values, name='ยอดขายรายเมนู'), row=1, col=2)
```

- `go.Bar`: ข้อมูลกราฟแท่ง
- `x / y`: ชื่อเมนูและยอดขาย
- `row=1, col=2`: ช่องกราฟแท่ง
- `add_trace(...)`: เพิ่มชุดข้อมูลกราฟแท่งในตำแหน่ง row และ col
- `name`: ชื่อชุดข้อมูลที่ใช้กำกับในกราฟ

ดาวน์โหลด W12_Dashboard.html จาก Files และเปิดในเบราว์เซอร์เพื่อตรวจตัวเลือกอีกครั้ง

**เมื่อเกิดข้อผิดพลาด:** ตรวจชื่อคอลัมน์ ตัวแปร และวงเล็บ เรียกใช้เซลล์ก่อนหน้าตามลำดับก่อนดำเนินการต่อ

**ตรวจความเข้าใจ:** อธิบายหน้าที่ของคำสั่ง และเปรียบเทียบผลลัพธ์กับตัวอย่าง


## สรุป

อธิบายคำสั่ง ผลลัพธ์ หน่วย และข้อจำกัด บันทึก Notebook สำหรับทบทวน
