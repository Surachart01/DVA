## ขั้นที่ 2: [Get Data / Extract] pd.read_csv() นำเข้าไฟล์ข้อมูลและกำหนดชนิดข้อมูลหมายเลขโทรศัพท์

ฟังก์ชัน pd.read_csv() ทำหน้าที่ดึงข้อมูล (Extract) จากไฟล์ CSV เข้าสู่ DataFrame โดยใช้พารามิเตอร์ dtype กำหนดให้คอลัมน์หมายเลขโทรศัพท์เป็นข้อความเพื่อรักษาเลข 0 นำหน้า

## ขั้นที่ 5: [Get Data] .shape ตรวจสอบขนาดของตาราง (จำนวนแถวและจำนวนคอลัมน์)

```python
df.shape
```

- `df`: ตัวแปรตารางข้อมูลต้นฉบับ
- `.shape`: คุณลักษณะแสดงขนาดตารางในรูปแบบ (แถว, คอลัมน์) โดยไม่ใช้วงเล็บ

```text
(36, 7)
```

**เมื่อเกิดข้อผิดพลาด:** ห้ามใส่เครื่องหมายวงเล็บเป็น df.shape() เนื่องจาก shape เป็นคุณลักษณะของตาราง ไม่ใช่เมธอด

**ตรวจความเข้าใจ:** ตัวเลข 36 และ 7 ในผลลัพธ์ (36, 7) มีความหมายถึงสิ่งใดตามลำดับ

## ขั้นที่ 7: [การเชื่อมโยงสู่ ETL] .info() ตรวจสอบโครงสร้าง ชนิดข้อมูล และจำนวนข้อมูลที่ไม่เป็นค่าว่าง

เมธอด .info() แสดงข้อมูลสรุปของตาราง ประกอบด้วยจำนวนแถว รายชื่อคอลัมน์ จำนวนข้อมูลที่ไม่เป็นค่าว่าง (Non-Null Count) และชนิดข้อมูล (Dtype)

```python
df.info()
```

- `df`: ตัวแปรตารางข้อมูลต้นฉบับ
- `.info()`: เมธอดแสดงรายงานโครงสร้างและจำนวนข้อมูลที่มีอยู่ในแต่ละคอลัมน์

เขียนคำสั่งและกด Shift + Enter ก่อนอ่านผลด้านล่าง


### การตรวจผล

```text
<class 'pandas.core.frame.DataFrame'>
RangeIndex: 36 entries, 0 to 35
Data columns (total 7 columns):
 #   Column          Non-Null Count  Dtype  
---  ------          --------------  -----  
 0   order_id        36 non-null     object 
 1   customer_phone  34 non-null     string 
 2   product_name    36 non-null     object 
 3   category        36 non-null     object 
 4   quantity        35 non-null     float64
 5   total_price     34 non-null     float64
 6   payment_status  36 non-null     object 
dtypes: float64(2), object(4), string(1)
memory usage: 2.1+ KB
```

จากข้อมูลทั้งหมด 36 แถว พบว่าคอลัมน์ customer_phone มีข้อมูล 34 แถว, quantity มีข้อมูล 35 แถว และ total_price มีข้อมูล 34 แถว แสดงว่าทั้ง 3 คอลัมน์นี้มีค่าว่างปรากฏอยู่

**เมื่อเกิดข้อผิดพลาด:** ผู้เรียนควรตรวจสอบค่าในคอลัมน์ Non-Null Count ร่วมกับ Dtype เสมอ เพื่อระบุคอลัมน์ที่มีข้อมูลสูญหาย

**ตรวจความเข้าใจ:** จากการตรวจสอบค่า Non-Null Count คอลัมน์ใดบ้างที่มีจำนวนข้อมูลไม่ครบ 36 แถว


## ขั้นที่ 8: [การเชื่อมโยงสู่ ETL] .describe() ตรวจสอบค่าทางสถิติพื้นฐานเพื่อค้นหาค่าผิดปกติ (Outlier)

เมธอด .describe() คำนวณค่าสถิติเชิงพรรณนา (count, mean, std, min, 25%, 50%, 75%, max) ของคอลัมน์ตัวเลข เพื่อตรวจสอบขอบเขตข้อมูล

```python
df.describe()
```

- `df`: ตัวแปรตารางข้อมูลต้นฉบับ
- `.describe()`: เมธอดคำนวณค่าสถิติพื้นฐานของคอลัมน์ที่มีชนิดข้อมูลเป็นตัวเลข

เขียนคำสั่งและกด Shift + Enter ก่อนอ่านผลด้านล่าง


### การตรวจผล

| ดัชนี | quantity | total_price |
| --- | --- | --- |
| count | 35.0 | 34.0 |
| mean | 1.6285714285714286 | 3613.823529411765 |
| std | 1.1653253594071105 | 16860.921829414056 |
| min | -2.0 | 180.0 |
| 25% | 1.0 | 395.0 |
| 50% | 1.0 | 620.0 |
| 75% | 2.0 | 957.5 |
| max | 5.0 | 99000.0 |

พบค่าผิดปกติ 2 รายการ ได้แก่ คอลัมน์ quantity มีค่าต่ำสุด (min) เท่ากับ -2 ชิ้น และคอลัมน์ total_price มีค่าสูงสุด (max) เท่ากับ 99,000 บาท ส่งผลให้ค่าเฉลี่ย (mean) สูงถึง 3,677.94 บาท

**เมื่อเกิดข้อผิดพลาด:** หากไม่ตรวจสอบค่าต่ำสุดและค่าสูงสุดด้วย .describe() ก่อนนำไปคำนวณค่าเฉลี่ย จะทำให้ได้ผลสรุปที่คลาดเคลื่อนจากข้อมูลจริง

**ตรวจความเข้าใจ:** ค่าต่ำสุด (min) ของ quantity และค่าสูงสุด (max) ของ total_price มีความผิดปกติในเชิงธุรกิจอย่างไร

## ขั้นที่ 15: [ETL: Transform 2] .drop_duplicates() ลบแถวข้อมูลที่ซ้ำซ้อนและแยกจัดเก็บในตัวแปร df_clean

- `df_clean.shape`: ตรวจสอบขนาดตารางภายหลังการลบแถวข้อมูลที่ซ้ำซ้อน

## ขั้นที่ 16: [ETL: Transform 3] .dropna(subset=...) คัดกรองแถวที่มีค่าว่างในคอลัมน์สำคัญออก

ใช้เมธอด .dropna() ร่วมกับพารามิเตอร์ subset เพื่อลบเฉพาะแถวที่ไม่มีข้อมูลจำนวนชิ้น (quantity) หรือยอดเงินรวม (total_price) โดยคงแถวที่ขาดเฉพาะหมายเลขโทรศัพท์ไว้

```python
df_clean = df_clean.dropna(subset=['quantity', 'total_price'])
df_clean.shape
```

- `df_clean.dropna(...)`: เมธอดสำหรับลบแถวที่ปรากฏค่าว่าง (Missing Values)
- `subset=['quantity', 'total_price']`: พารามิเตอร์กำหนดให้ตรวจสอบค่าว่างเฉพาะในคอลัมน์ quantity และ total_price
- `df_clean.shape`: ตรวจสอบขนาดตารางภายหลังการคัดกรองแถวที่มีค่าว่าง

เขียนคำสั่งและกด Shift + Enter ก่อนอ่านผลด้านล่าง


### การตรวจผล

```text
(31, 7)
```

ขนาดของตารางลดลงจาก (34, 7) เหลือ (31, 7) โดยตัดเฉพาะ 3 แถวที่ไม่สามารถคำนวณมูลค่าการซื้อขายได้ ในขณะที่รายการที่ขาดเฉพาะหมายเลขโทรศัพท์ยังคงอยู่ในระบบ

**เมื่อเกิดข้อผิดพลาด:** หากเรียกใช้ df_clean.dropna() โดยไม่ระบุพารามิเตอร์ subset ระบบจะลบรายการ ORD-104 และ ORD-122 ซึ่งมีมูลค่าการซื้อขายจริงแต่ไม่ระบุหมายเลขโทรศัพท์ออกไปด้วย

**ตรวจความเข้าใจ:** การกำหนดพารามิเตอร์ subset=['quantity', 'total_price'] มีข้อดีอย่างไรเมื่อเปรียบเทียบกับการใช้คำสั่ง dropna() ทั้งตาราง


## ขั้นที่ 17: [ETL: Transform 4] .between() คัดกรองข้อมูลให้อยู่ในเกณฑ์ที่กำหนดและเปรียบเทียบค่าเฉลี่ย

ใช้เมธอด .between(1, 20) เพื่อคัดกรองจำนวนชิ้นที่อยู่ในช่วง 1 ถึง 20 ชิ้น และคัดกรองยอดเงินรวมไม่เกิน 10,000 บาท จากนั้นเปรียบเทียบค่าเฉลี่ยก่อนและหลังการจัดเตรียมข้อมูล

```python
df_clean = df_clean[df_clean['quantity'].between(1, 20) & (df_clean['total_price'] <= 10000)]
round(df['total_price'].mean(), 2), round(df_clean['total_price'].mean(), 2)
```

- `df_clean['quantity'].between(1, 20)`: เมธอดตรวจสอบว่าค่าในคอลัมน์ quantity อยู่ในช่วงตั้งแต่ 1 ถึง 20 หรือไม่
- `& (df_clean['total_price'] <= 10000)`: ตัวดำเนินการเชื่อมเงื่อนไขว่ายอดเงินรวมต่อรายการต้องไม่เกิน 10,000 บาท
- `round(..., 2)`: ฟังก์ชันปัดเศษทศนิยม 2 ตำแหน่งเพื่อเปรียบเทียบค่าเฉลี่ยของตาราง df และ df_clean

## ขั้นที่ 18: [ETL: Load] .to_csv() บันทึกตารางข้อมูลที่ผ่านการจัดเตรียมออกเป็นไฟล์ CSV ใหม่

ขั้นตอนสุดท้ายของกระบวนการ ETL (Load): ใช้เมธอด .to_csv() เพื่อส่งออกตารางข้อมูล df_clean (จำนวน 29 แถว) เป็นไฟล์ CSV ใหม่โดยไม่บันทึกหมายเลขดัชนี

- `df_clean.to_csv('nvc_tech_store_clean.csv', ...)`: เมธอดสำหรับส่งออกตารางข้อมูลเป็นไฟล์ CSV ตามชื่อที่กำหนด
- `index=False`: พารามิเตอร์กำหนดไม่ให้บันทึกหมายเลขดัชนีของตารางเป็นคอลัมน์เพิ่มเติมในไฟล์ CSV
- `df_clean.head()`: แสดงข้อมูลตัวอย่าง 5 แถวแรกของตารางที่ผ่านการจัดเตรียมเรียบร้อยแล้ว
