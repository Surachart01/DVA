## ขั้นตอนที่ 1: การนำเข้าไลบรารี pandas

นำเข้าไลบรารี pandas และกำหนดชื่อย่อเป็น pd สำหรับใช้คำสั่งจัดการตารางข้อมูลและสร้างตารางรายงานคุณภาพข้อมูล

```python
import pandas as pd
```

- `import pandas`: คำสั่งนำเข้าไลบรารี pandas เข้าสู่สภาพแวดล้อมการทำงาน
- `as pd`: กำหนดชื่อย่อเป็น pd เพื่อความสะดวกในการเรียกใช้ฟังก์ชัน

**ข้อควรระวังในการเขียนคำสั่ง:** ต้องพิมพ์คำว่า import, pandas และ pd ด้วยตัวอักษรพิมพ์เล็กทั้งหมด และต้องรันเซลล์นี้ทุกครั้งเมื่อเริ่มต้น Runtime ใหม่

## ขั้นตอนที่ 2: อ่านไฟล์ข้อมูลและแสดงตัวอย่าง 5 แถวแรก

อ่านไฟล์ student_register_dirty.csv เข้าสู่ตัวแปร df โดยให้คอลัมน์ phone เป็นข้อความ (string) เพื่อไม่ให้เลข 0 นำหน้าหาย แล้วแสดงตัวอย่าง 5 แถวแรก

```python
df = pd.read_csv('student_register_dirty.csv', dtype={'phone': 'string'})
df.head()
```

- `pd.read_csv('student_register_dirty.csv', ...)`: อ่านไฟล์ CSV เข้าสู่ตาราง DataFrame ชื่อ df
- `dtype={'phone': 'string'}`: ให้คอลัมน์ phone เป็นข้อความ เพื่อไม่ให้เลข 0 นำหน้าหาย
- `df.head()`: แสดงตัวอย่าง 5 แถวแรก (ดัชนี 0–4) โดยไม่ได้ตัดข้อมูลใน df

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์

## ขั้นตอนที่ 3: การกำหนดตัวหาร (Denominator) จากจำนวนแถวทั้งหมด

คำนวณจำนวนแถวทั้งหมดของตารางข้อมูลต้นฉบับด้วยฟังก์ชัน len(df) และเก็บไว้ในตัวแปร row_count เพื่อใช้เป็นตัวหารมาตรฐานในการคำนวณร้อยละ

```python
row_count = len(df)
row_count
```

- `len(df)`: ฟังก์ชันนับจำนวนแถวทั้งหมดในตาราง DataFrame
- `row_count =`: เก็บค่าจำนวนแถวทั้งหมดไว้ในตัวแปรเพื่อใช้เป็นตัวหาร

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์

**ข้อควรระวังในการเขียนคำสั่ง:** ต้องระบุในรายงานอย่างชัดเจนว่าตัวหาร 42 มาจากจำนวนแถวทั้งหมดในไฟล์ต้นฉบับ ไม่ใช่จำนวนผู้เรียนที่ไม่ซ้ำกัน

## ขั้นตอนที่ 4: การสร้างตารางตรวจสอบตำแหน่งข้อมูลสูญหาย (Missing Values)

ตรวจสอบตำแหน่งที่มีค่าว่างในทุกช่องของตาราง df ด้วยคำสั่ง .isna() เก็บในตัวแปร missing แล้วแสดงตัวอย่าง 5 แถวแรก

```python
missing = df.isna()
missing.head()
```

- `df.isna()`: ตรวจสอบทุกช่องในตาราง หากเป็นค่าว่างจะให้ค่า True หากมีข้อมูลจะให้ค่า False
- `missing.head()`: แสดงตัวอย่างตารางผลลัพธ์การตรวจค่าว่าง 5 แถวแรก

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์

| --- | --- | --- | --- | --- | --- | --- | --- |
| 0 | False | False | False | False | False | False | False |
| 1 | False | False | False | False | False | False | False |
| 2 | False | False | False | False | False | False | False |
| 3 | False | False | False | False | True | False | False |
| 4 | False | False | False | False | False | False | False |

ปรากฏค่า True ที่คอลัมน์ gpa ในแถวดัชนีที่ 3 (รหัส S004) ส่วนช่องที่มีข้อมูลปกติจะแสดงค่า False โดยตัวแปร missing มีขนาด 42 แถวเท่าตารางต้นฉบับ

**ข้อควรระวังในการเขียนคำสั่ง:** ค่า True ในตาราง missing หมายถึง 'พบค่าว่าง (ข้อมูลสูญหาย)' ไม่ได้หมายถึงข้อมูลถูกต้อง

**คำถามตรวจสอบความเข้าใจ:** อธิบายความหมายของค่า True และ False ในตาราง missing ได้ถูกต้อง


## ขั้นตอนที่ 5: การคำนวณจำนวนข้อมูลสูญหายแยกตามรายคอลัมน์

รวมจำนวนค่าว่าง (ค่า True มีค่าเท่ากับ 1) แยกตามรายคอลัมน์ และเก็บผลลัพธ์ไว้ในตัวแปร Series ชื่อ missing_count

```python
missing_count = missing.sum()
missing_count
```

- `missing.sum()`: นับผลรวมของค่า True แยกตามแต่ละคอลัมน์
- `missing_count`: ตัวแปร Series ที่เก็บจำนวนค่าว่างของทั้ง 7 คอลัมน์

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์


### ผลลัพธ์และคำอธิบาย

```text
student_id      0
name            0
major           0
age             1
gpa             2
phone           1
updated_date    0
```

พบค่าว่างใน 3 คอลัมน์ ได้แก่ age จำนวน 1 แถว, gpa จำนวน 2 แถว และ phone จำนวน 1 แถว ส่วนคอลัมน์อื่นมีค่าว่างเป็น 0

**ข้อควรระวังในการเขียนคำสั่ง:** ห้ามนำจำนวนค่าว่างของแต่ละคอลัมน์มาบวกรวมกันเป็นจำนวนผู้เรียนทันที เนื่องจากผู้เรียน 1 คนอาจมีค่าว่างมากกว่า 1 คอลัมน์ได้

**คำถามตรวจสอบความเข้าใจ:** ระบุคอลัมน์ที่พบค่าว่างทั้ง 3 คอลัมน์พร้อมจำนวนที่พบได้ถูกต้อง


## ขั้นตอนที่ 6: การคำนวณร้อยละของข้อมูลสูญหายเทียบกับตัวหาร

นำจำนวนค่าว่างของแต่ละคอลัมน์ (missing_count) มาหารด้วยจำนวนแถวทั้งหมด (row_count) แล้วคูณด้วย 100 เพื่อแปลงเป็นค่าร้อยละ

```python
missing_percent = missing_count / row_count * 100
missing_percent
```

- `missing_percent =`: สร้างตัวแปร missing_percent เพื่อเก็บร้อยละของค่าว่างแต่ละคอลัมน์ไว้ใช้สร้างตารางรายงาน
- `missing_count / row_count`: คำนวณสัดส่วนของค่าว่างต่อจำนวนแถวทั้งหมด (42 แถว)
- `* 100`: แปลงค่าสัดส่วนให้เป็นหน่วยร้อยละ (%)

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์


### ผลลัพธ์และคำอธิบาย

```text
student_id      0.000000
name            0.000000
major           0.000000
age             2.380952
gpa             4.761905
phone           2.380952
updated_date    0.000000
```

คอลัมน์ age และ phone มีค่าว่างคิดเป็นร้อยละ 2.380952 (จาก 1 ÷ 42 × 100) และคอลัมน์ gpa มีค่าว่างคิดเป็นร้อยละ 4.761905 (จาก 2 ÷ 42 × 100)

**ข้อควรระวังในการเขียนคำสั่ง:** ระวังไม่คูณ 100 ซ้ำซ้อน และควรเก็บค่าทศนิยมจริงไว้ในตารางรายงานก่อนนำไปสรุปผล

**คำถามตรวจสอบความเข้าใจ:** อธิบายที่มาของค่าร้อยละ 4.761905 ในคอลัมน์ gpa ด้วยตัวตั้งและตัวหารได้


## ขั้นตอนที่ 7: การสร้างตารางรายงานค่าว่างด้วย pd.DataFrame()

นำตัวแปร Series ทั้งสองตัว (จำนวนค่าว่างและร้อยละของค่าว่าง) มารวมกันเป็นตารางรายงาน DataFrame ชื่อ quality

```python
quality = pd.DataFrame({'missing_count': missing_count, 'missing_percent': missing_percent})
quality
```

- `quality =`: สร้างตัวแปร quality เพื่อเก็บตารางรายงานที่ได้
- `pd.DataFrame({...})`: สร้างตาราง DataFrame จากโครงสร้างพจนานุกรม (Dictionary)
- `'missing_count': missing_count`: กำหนดคอลัมน์แรกให้แสดงจำนวนค่าว่าง
- `'missing_percent': missing_percent`: กำหนดคอลัมน์ที่สองให้แสดงร้อยละของค่าว่าง

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์


### ผลลัพธ์และคำอธิบาย

| ลำดับแถว | missing_count | missing_percent |
| --- | --- | --- |
| student_id | 0 | 0.0 |
| name | 0 | 0.0 |
| major | 0 | 0.0 |
| age | 1 | 2.380952380952381 |
| gpa | 2 | 4.761904761904762 |
| phone | 1 | 2.380952380952381 |
| updated_date | 0 | 0.0 |

ได้ตารางรายงานขนาด 7 แถว 2 คอลัมน์ โดยมีรายชื่อคอลัมน์ต้นทาง (student_id ถึง updated_date) ทำหน้าที่เป็นดัชนีแถว (Index) ด้านซ้ายมือ

**ข้อควรระวังในการเขียนคำสั่ง:** สังเกตว่าชื่อคอลัมน์ต้นทางอยู่ในส่วนดัชนี (Index) ไม่ใช่คอลัมน์ปกติ จึงต้องระมัดระวังในการส่งออกเป็นไฟล์ CSV ในขั้นตอนถัดไป

**คำถามตรวจสอบความเข้าใจ:** ตรวจสอบตารางผลลัพธ์ว่าแถว gpa แสดงค่า missing_count เท่ากับ 2 และ missing_percent ประมาณ 4.761905


## ขั้นตอนที่ 8: การบันทึกตารางรายงานค่าว่างเป็นไฟล์ CSV พร้อมกำหนดชื่อดัชนี

ส่งออกตารางรายงาน quality เป็นไฟล์ CSV ชื่อ quality_report.csv โดยใช้พารามิเตอร์ index_label='column' เพื่อตั้งชื่อหัวคอลัมน์ให้แก่ดัชนี

```python
quality.to_csv('quality_report.csv', index_label='column', encoding='utf-8-sig')
```

- `quality.to_csv('quality_report.csv', ...)`: บันทึกตาราง quality ออกเป็นไฟล์ CSV
- `index_label='column'`: เก็บรักษาดัชนี (รายชื่อคอลัมน์ต้นทาง) ไว้และตั้งชื่อหัวคอลัมน์ว่า column
- `encoding='utf-8-sig'`: กำหนดรหัสอักขระเพื่อให้เปิดไฟล์ในโปรแกรม Microsoft Excel แล้วอ่านภาษาไทยได้ถูกต้อง

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์


### ผลลัพธ์และคำอธิบาย

```text
รันคำสั่งสำเร็จ (ไม่มีข้อความแจ้งข้อผิดพลาด)
```

คำสั่งทำงานสำเร็จและสร้างไฟล์ quality_report.csv ในแถบ Files โดยประกอบด้วย 3 คอลัมน์ ได้แก่ column, missing_count และ missing_percent

**ข้อควรระวังในการเขียนคำสั่ง:** ห้ามระบุ index=False ในขั้นตอนนี้โดยเด็ดขาด เพราะจะทำให้รายชื่อคอลัมน์ต้นทาง (student_id, age, gpa ฯลฯ) ถูกลบหายไปจากไฟล์รายงาน

**คำถามตรวจสอบความเข้าใจ:** ตรวจสอบในแถบ Files ของ Google Colab ว่าปรากฏไฟล์ชื่อ quality_report.csv


## ขั้นตอนที่ 9: การอ่านไฟล์รายงาน CSV กลับมาตรวจสอบความสมบูรณ์

## ขั้นตอนที่ 11: นับชื่อสาขาวิชาที่ไม่ตรงมาตรฐาน

สร้างรายชื่อสาขาวิชามาตรฐาน แล้วแสดงแถวที่ชื่อสาขาวิชาไม่อยู่ในรายชื่อนั้น

```python
major_ok = df['major'].isin(['เทคโนโลยีสารสนเทศ', 'คอมพิวเตอร์ธุรกิจ', 'การตลาด'])
major_bad = ~major_ok
df.loc[major_bad, ['student_id', 'major']]
```

- `major_ok =`: สร้างตัวแปร major_ok เก็บผล True/False ว่าแต่ละแถวใช้ชื่อสาขาวิชามาตรฐานหรือไม่
- `['เทคโนโลยีสารสนเทศ', 'คอมพิวเตอร์ธุรกิจ', 'การตลาด']`: รายชื่อสาขาวิชามาตรฐาน (list) ที่ยอมรับในรายงานนี้
- `.isin([...])`: ตรวจว่าค่าในแต่ละแถวอยู่ในรายชื่อที่กำหนดหรือไม่ ได้ True เมื่ออยู่ในรายชื่อ และ False เมื่อไม่อยู่
- `major_bad = ~major_ok`: เครื่องหมาย ~ กลับค่า True เป็น False และ False เป็น True จึงได้ True เฉพาะแถวที่ชื่อสาขาวิชาไม่ตรงมาตรฐาน
- `df.loc[major_bad, ['student_id', 'major']]`: df.loc[แถว, คอลัมน์] เลือกเฉพาะแถวที่ major_bad เป็น True และแสดงเฉพาะคอลัมน์ student_id กับ major

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์

| ลำดับแถว | student_id | major |
| --- | --- | --- |
| 12 | S013 | IT |
| 27 | S028 | เทคโนโลยี สารสนเทศ |

พบ 2 แถว ได้แก่ รหัส S013 (IT) และรหัส S028 (เทคโนโลยี สารสนเทศ)

**ข้อควรระวังในการเขียนคำสั่ง:** ต้องสะกดชื่อสาขาวิชามาตรฐานในรายการให้ตรงทุกตัวอักษร หากพิมพ์ผิด แถวที่ถูกต้องจะถูกนับเป็นปัญหาด้วย

**คำถามตรวจสอบความเข้าใจ:** เหตุใดจึงใช้ .isin() กับรายชื่อมาตรฐาน แทนการค้นหาชื่อที่ผิดทีละชื่อ


## ขั้นตอนที่ 12: ตรวจจำนวนหลักของหมายเลขโทรศัพท์

นับจำนวนตัวอักษรของหมายเลขโทรศัพท์ แล้วแสดงแถวที่ไม่ครบ 10 หลัก โดยไม่นับค่าว่าง

```python
phone_bad = (df['phone'].str.len() != 10) & df['phone'].notna()
df.loc[phone_bad, ['student_id', 'phone']]
```

- `phone_bad =`: สร้างตัวแปร phone_bad เก็บผล True/False ว่าแต่ละแถวมีหมายเลขโทรศัพท์ไม่ครบ 10 หลักหรือไม่
- `df['phone'].str`: เรียกใช้คำสั่งกลุ่มข้อความ (.str) กับคอลัมน์ phone ทุกแถวพร้อมกัน
- `.str.len()`: นับจำนวนตัวอักษรของหมายเลขโทรศัพท์แต่ละแถว เช่น 0882903368 ได้ 10
- `!= 10`: เปรียบเทียบว่าไม่เท่ากับ 10 ได้ True เมื่อจำนวนหลักไม่ครบหรือเกิน 10 หลัก
- `& df['phone'].notna()`: .notna() ได้ True เมื่อช่องนั้นมีข้อมูล ส่วน & หมายถึงต้องเป็นจริงทั้งสองเงื่อนไข จึงไม่นับค่าว่างซึ่งรายงานไว้แล้วในขั้นตอนที่ 4–6
- `df.loc[phone_bad, ['student_id', 'phone']]`: เลือกเฉพาะแถวที่ phone_bad เป็น True และแสดงรหัสผู้เรียนกับหมายเลขโทรศัพท์ เพื่อใช้เป็นหลักฐานในรายงาน

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์

## ขั้นตอนที่ 13: ตรวจความเป็นปัจจุบันของข้อมูล

แปลงคอลัมน์วันที่ให้เป็นชนิดวันที่ แล้วแสดงแถวที่ปรับปรุงข้อมูลก่อนวันที่ 1 มกราคม 2026 ซึ่งเป็นเกณฑ์ความเป็นปัจจุบันของรายงานนี้

```python
updated = pd.to_datetime(df['updated_date'], format='%Y-%m-%d')
old_record = updated < '2026-01-01'
df.loc[old_record, ['student_id', 'updated_date']]
```

- `updated =`: สร้างตัวแปร updated เก็บวันที่ปรับปรุงข้อมูลที่แปลงเป็นชนิดวันที่แล้ว
- `pd.to_datetime(df['updated_date'], ...)`: แปลงข้อความวันที่ให้เป็นชนิดวันที่ (datetime) ทำให้เปรียบเทียบก่อน-หลังได้ถูกต้อง
- `format='%Y-%m-%d'`: บอกรูปแบบของข้อความวันที่ว่าเป็น ปี 4 หลัก-เดือน-วัน เช่น 2026-06-01
- `old_record = updated < '2026-01-01'`: เปรียบเทียบว่าวันที่เก่ากว่า 1 มกราคม 2026 หรือไม่ ได้ True เมื่อข้อมูลไม่เป็นปัจจุบันตามเกณฑ์
- `df.loc[old_record, ['student_id', 'updated_date']]`: แสดงเฉพาะแถวที่ข้อมูลไม่เป็นปัจจุบัน พร้อมรหัสผู้เรียนและวันที่ปรับปรุง

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์

## ขั้นตอนที่ 14: นับแถวที่มีค่าว่างอย่างน้อย 1 ช่อง

ตรวจทีละแถวว่ามีค่าว่างอย่างน้อย 1 ช่องหรือไม่ แล้วนับจำนวนแถวที่ข้อมูลไม่สมบูรณ์

```python
missing_rows = df.isna().any(axis=1)
missing_rows.sum()
```

- `missing_rows =`: สร้างตัวแปร missing_rows เก็บผล True/False ว่าแต่ละแถวมีค่าว่างหรือไม่
- `df.isna()`: สร้างตาราง True/False โดยช่องที่เป็นค่าว่างได้ True
- `.any(axis=1)`: .any() ตรวจว่ามี True อย่างน้อย 1 ค่าหรือไม่ ส่วน axis=1 หมายถึงตรวจตามแนวแถว (จากซ้ายไปขวา) จึงได้ผล 1 ค่าต่อ 1 แถว
- `missing_rows.sum()`: นับจำนวนแถวที่เป็น True คือจำนวนแถวที่ข้อมูลไม่สมบูรณ์

## ขั้นตอนที่ 15: สร้างรายงานตามมิติคุณภาพข้อมูล

รวมผลการตรวจทั้ง 4 มิติไว้ในตารางเดียว โดยระบุมิติ ปัญหา จำนวน และตัวหาร

```python
dimension_report = pd.DataFrame({
    'dimension': ['ความสมบูรณ์', 'ความสม่ำเสมอ', 'ความถูกต้อง', 'ความเป็นปัจจุบัน'],
    'issue': ['แถวที่มีค่าว่างอย่างน้อย 1 ช่อง', 'ชื่อสาขาวิชาไม่ตรงมาตรฐาน', 'หมายเลขโทรศัพท์ไม่ครบ 10 หลัก', 'ปรับปรุงข้อมูลก่อนปี 2026'],
    'count': [missing_rows.sum(), major_bad.sum(), phone_bad.sum(), old_record.sum()],
    'denominator': row_count
})
dimension_report
```

- `dimension_report = pd.DataFrame({...})`: pd.DataFrame() สร้างตารางใหม่จากข้อมูลในรูป {ชื่อคอลัมน์: รายการค่า} แล้วเก็บไว้ในตัวแปร dimension_report
- `'dimension': [...]`: คอลัมน์ชื่อมิติคุณภาพข้อมูล 4 มิติ
- `'issue': [...]`: คอลัมน์คำอธิบายปัญหาที่ตรวจพบในแต่ละมิติ
- `'count': [missing_rows.sum(), ...]`: คอลัมน์จำนวนแถวที่พบปัญหา โดยใช้ .sum() นับ True จากตัวแปรในขั้นตอนที่ 11–14
- `'denominator': row_count`: คอลัมน์ตัวหาร ใส่ค่าเดียว (42) แล้ว pandas จะใส่ให้ทุกแถวโดยอัตโนมัติ

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์


### ผลลัพธ์และคำอธิบาย

| ลำดับแถว | dimension | issue | count | denominator |
| --- | --- | --- | --- | --- |
| 0 | ความสมบูรณ์ | แถวที่มีค่าว่างอย่างน้อย 1 ช่อง | 4 | 42 |
| 1 | ความสม่ำเสมอ | ชื่อสาขาวิชาไม่ตรงมาตรฐาน | 2 | 42 |
| 2 | ความถูกต้อง | หมายเลขโทรศัพท์ไม่ครบ 10 หลัก | 1 | 42 |
| 3 | ความเป็นปัจจุบัน | ปรับปรุงข้อมูลก่อนปี 2026 | 2 | 42 |

ได้ตาราง 4 แถว 4 คอลัมน์ จำนวนที่พบคือ 4, 2, 1 และ 2 แถว ตามลำดับ โดยมีตัวหาร 42 ทุกแถว

**ข้อควรระวังในการเขียนคำสั่ง:** ลำดับรายการในทุกคอลัมน์ต้องตรงกัน และแต่ละรายการต้องมีจำนวนข้อมูลเท่ากัน มิฉะนั้นจะเกิดข้อผิดพลาด ValueError

**คำถามตรวจสอบความเข้าใจ:** มิติใดพบปัญหามากที่สุด


## ขั้นตอนที่ 16: คำนวณร้อยละและบันทึกรายงานตามมิติ

เพิ่มคอลัมน์ร้อยละโดยปัดเป็นทศนิยม 2 ตำแหน่ง แล้วบันทึกรายงานเป็นไฟล์ CSV

```python
dimension_report['percent'] = (dimension_report['count'] / dimension_report['denominator'] * 100).round(2)
dimension_report.to_csv('quality_dimensions.csv', index=False, encoding='utf-8-sig')
dimension_report
```

- `dimension_report['percent'] =`: สร้างคอลัมน์ใหม่ชื่อ percent ในตาราง dimension_report
- `dimension_report['count'] / dimension_report['denominator'] * 100`: คำนวณร้อยละทีละแถว จากจำนวนที่พบหารด้วยตัวหาร แล้วคูณ 100
- `.round(2)`: ปัดตัวเลขเป็นทศนิยม 2 ตำแหน่ง เพื่อให้อ่านรายงานได้ง่าย
- `to_csv('quality_dimensions.csv', index=False, ...)`: บันทึกตารางเป็นไฟล์ CSV โดย index=False ไม่บันทึกเลขลำดับแถว เพราะมีคอลัมน์ dimension ระบุรายการอยู่แล้ว
- `encoding='utf-8-sig'`: กำหนดรหัสอักขระ ให้เปิดไฟล์ใน Excel แล้วอ่านภาษาไทยได้ถูกต้อง

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์
