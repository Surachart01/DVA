## ขั้นตอนที่ 4: แปลงวันที่เป็น datetime

แปลงข้อความวันที่เป็นชนิดวันที่แล้วเก็บกลับคอลัมน์เดิม

```python
df['date'] = pd.to_datetime(df['date'], format='%Y-%m-%d', errors='raise')
```

- `df['date'] =`: แทนคอลัมน์ date ด้วยผลแปลง
- `pd.to_datetime(...)`: แปลงข้อมูลเป็น datetime
- `df['date']`: ข้อมูลวันที่ที่จะนำมาแปลง
- `format='%Y-%m-%d'`: %Y คือปี ค.ศ. สี่หลัก %m คือเดือน %d คือวันที่ คั่นด้วยขีดกลาง
- `errors='raise'`: ให้เกิดข้อผิดพลาดเมื่อพบวันที่แปลงไม่ได้

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์


````python
df['date'] = pd.to_datetime(df['date'], format='%Y-%m-%d', errors='raise')
````

### ผลลัพธ์และคำอธิบาย

```text

```

เก็บวันที่ชนิด datetime แล้ว โดยยังไม่แสดงตาราง

**ข้อควรระวังในการเขียนคำสั่ง:** ชุดพร้อมสอนเป็น ค.ศ. อยู่แล้ว ห้ามลบ 543 ซ้ำ

**คำถามตรวจสอบความเข้าใจ:** รันได้โดยไม่เกิดข้อผิดพลาดและอธิบายรูปแบบวันได้

## ขั้นตอนที่ 12: กำหนดหน้าต่างสามรายการ

กำหนดการพิจารณาสามรายการย้อนหลังรวมรายการปัจจุบัน

```python
revenue_window = daily['revenue'].rolling(window=3, min_periods=3)
```

- `revenue_window =`: เก็บหน้าต่างข้อมูลเพื่อคำนวณ
- `daily['revenue']`: เลือกยอดเงินที่เรียงวันแล้ว
- `window=3`: ใช้สามรายการล่าสุดในแต่ละตำแหน่ง
- `min_periods=3`: ต้องมีค่าจริงครบสามรายการจึงให้ผล

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์


````python
revenue_window = daily['revenue'].rolling(window=3, min_periods=3)
````

### ผลลัพธ์และคำอธิบาย

```text
รันคำสั่งสำเร็จ (ไม่มีข้อความแจ้งข้อผิดพลาด)
```

ยังไม่คำนวณค่าเฉลี่ยและยังไม่แสดงตาราง

**ข้อควรระวังในการเขียนคำสั่ง:** สามรายการไม่ใช่สามวันต่อเนื่องเมื่อมีวันขาด

**คำถามตรวจสอบความเข้าใจ:** รันได้และอธิบายหน้าที่ window กับ min_periods
