## ขั้นที่ 7: เชื่อมรายการขายกับสินค้า

เติมชื่อสินค้าและราคาลงในแต่ละรายการขาย

```python
merged = sales.merge(products, on='product_id', how='left', validate='many_to_one')
```

- `sales.merge(products, ...)`: ใช้ sales เป็นตารางซ้ายและ products เป็นตารางขวา
- `on='product_id'`: จับคู่ด้วยรหัสสินค้าที่ชื่อเหมือนกันทั้งสองตาราง
- `how='left'`: เก็บทุกรายการจาก sales แม้หารหัสคู่ไม่พบ
- `validate='many_to_one'`: ตรวจว่าหลายรายการขายจับคู่กับสินค้าเพียงหนึ่งแถวต่อรหัส

เขียนคำสั่งและกด Shift + Enter ก่อนอ่านผลด้านล่าง


````python
merged = sales.merge(products, on='product_id', how='left', validate='many_to_one')
````

### การตรวจผล

**เมื่อเกิดข้อผิดพลาด:** อย่าใช้ inner เพื่อซ่อนรายการที่ไม่มีคู่; ตรวจชื่อคีย์และคีย์ซ้ำเมื่อเชื่อมไม่ผ่าน

**ตรวจความเข้าใจ:** รันได้โดยไม่เกิดข้อผิดพลาด แล้วแสดง merged ในขั้นถัดไป

- `index='sale_date'`: แถวคือวันที่ขาย

- `fill_value=0`: แทนช่องคู่วันกับหมวดที่ไม่มีรายการด้วยศูนย์
