## วันจันทร์ — วันจันทร์: เตรียมแฟ้มและนำเสนอโครงงานชุดแรก

- ตรวจความพร้อมของแฟ้มโครงงานและอุปกรณ์นำเสนอ
- ตรวจ Pipeline ทั้ง 4 ขั้นพร้อมหลักฐาน
- ซ้อมการนำเสนอและแบ่งบทบาทสมาชิก
- กลุ่มชุดแรกนำเสนอวันจันทร์ กลุ่มที่เหลือนำเสนอวันพฤหัสบดี


## วันจันทร์ — จัดแฟ้มโครงงานให้ตรวจสอบได้

- ไฟล์ข้อมูลที่เผยแพร่ได้ พร้อมที่มาและคำอธิบายคอลัมน์
- Notebook ที่รันตามลำดับและตรวจย้อนกลับได้
- Dashboard และสไลด์ที่ใช้ตัวเลขตรงกับ Notebook
- บันทึกบทบาทสมาชิกและรายการไฟล์หรือลิงก์ที่ส่ง


## วันจันทร์ — ตรวจหลักฐาน Pipeline ทั้ง 4 ขั้น

- Get Data: แหล่งข้อมูล คำถาม และความหมายคอลัมน์
- ETL: ปัญหาที่พบ วิธีจัดการ และผลก่อน–หลัง
- Modeling: คีย์ ความสัมพันธ์ สูตรคำนวณ และตัวชี้วัด
- Visualization: กราฟ Dashboard และข้อเสนอที่มีหลักฐาน


## วันจันทร์ — เกณฑ์โครงงาน 20 คะแนน

- ข้อมูล คุณภาพ และความปลอดภัย 4 คะแนน
- แบบจำลอง การคำนวณ และการตรวจผล 4 คะแนน
- Dashboard และหลักฐาน UX 4 คะแนน
- Insight การตัดสินใจ และข้อจำกัด 4 คะแนน
- การนำเสนอ ตอบคำถาม และบทบาทสมาชิก 4 คะแนน


## วันจันทร์ — ซ้อมการนำเสนอ 12 นาทีและการตอบคำถาม 5 นาที

- เริ่มจากคำถามและบริบทของข้อมูล
- แสดงวิธีเตรียมข้อมูลและผลสำคัญที่ตรวจย้อนกลับได้
- สาธิต Dashboard และเสนอการตัดสินใจพร้อมข้อจำกัด
- แบ่งผู้พูดและซ้อมเปิดหลักฐานสำหรับตอบคำถาม


## วันจันทร์ — นำเสนอกลุ่มชุดแรกและบันทึกข้อเสนอแนะ

- นำเสนอกลุ่มละ 12 นาที และตอบคำถาม 5 นาที
- สมาชิกอธิบายส่วนงานที่ตนรับผิดชอบพร้อมหลักฐาน
- ผู้ฟังบันทึกข้อสังเกตที่เกี่ยวกับข้อมูลและวิธีคิด
- ผู้สอนบันทึกผลตามเกณฑ์โครงงานและประเด็นที่ต้องตรวจเพิ่มเติม


## วันพฤหัสบดี — วันพฤหัสบดี: ตรวจความพร้อมของกลุ่มที่เหลือ

- ตรวจลำดับกลุ่มที่ยังไม่ได้นำเสนอ
- เปิดแฟ้ม Dashboard และสไลด์ก่อนถึงรอบ
- ยืนยันบทบาทผู้พูดและผู้เปิดหลักฐาน
- ใช้กรอบเวลาและเกณฑ์เดียวกับวันจันทร์


## วันพฤหัสบดี — นำเสนอกลุ่มที่เหลือและตอบคำถามรายบุคคล

- นำเสนอ 12 นาที และตอบคำถาม 5 นาทีต่อกลุ่ม
- สาธิตผลจากแฟ้มโครงงานฉบับเดียวกับที่ส่ง
- ชี้หลักฐานของสูตรคำนวณ วิธีจัดเตรียมข้อมูล และข้อจำกัด
- สมาชิกตอบในส่วนงานของตนและเชื่อมกับภาพรวมโครงงาน


## วันพฤหัสบดี — ตรวจแฟ้มฉบับส่งหลังการนำเสนอ

- ตรวจไฟล์ข้อมูล Notebook Dashboard สไลด์ และบันทึกบทบาทให้ครบ
- ตรวจลิงก์และสิทธิ์เข้าถึงตามช่องทางที่ผู้สอนกำหนด
- ระบุเวอร์ชันที่ส่งและรายการที่แก้ไขตามข้อเสนอแนะ
- เก็บหลักฐานการประเมินโครงงานในหมวดเดียว


## วันพฤหัสบดี — สะท้อนผลโครงงานและการทำงานร่วมกัน

- ระบุสิ่งที่ Pipeline ของกลุ่มทำได้พร้อมหลักฐาน
- อธิบายข้อจำกัดและสิ่งที่จะปรับปรุงหากมีข้อมูลเพิ่ม
- ทบทวนบทบาทที่สมาชิกแต่ละคนได้ปฏิบัติ
- คะแนนโครงงานเต็ม 20 คะแนน ไม่มีคะแนน Lab เพิ่ม
