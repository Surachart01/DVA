# ส่วนที่ 2: ทฤษฎีและหลักการจัดเตรียมข้อมูล (ETL)

## ความหมายและองค์ประกอบของกระบวนการ ETL (Extract – Transform – Load)
- Extract (การดึงและนำเข้าข้อมูล): ขั้นตอนการรวบรวมและอ่านข้อมูลจากแหล่งข้อมูลต้นทาง เช่น ไฟล์ CSV หรือไฟล์ Excel เข้าสู่โปรแกรมประมวลผล พร้อมกำหนดชนิดข้อมูลเบื้องต้นให้ถูกต้องตามลักษณะการใช้งาน
- Transform (การปรับรูปแบบและทำความสะอาดข้อมูล): ขั้นตอนการตรวจสอบ คัดกรอง และแก้ไขความผิดพลาดของข้อมูล เช่น การตัดอักขระช่องว่างส่วนเกิน การลบข้อมูลที่บันทึกซ้ำซ้อน การจัดการค่าว่าง และการคัดกรองค่าที่อยู่นอกเกณฑ์ที่กำหนด
- Load (การจัดเก็บและส่งออกข้อมูล): ขั้นตอนการบันทึกตารางข้อมูลที่ผ่านการจัดเตรียมแล้วออกเป็นไฟล์ข้อมูลชุดใหม่ที่มีความถูกต้องสมบูรณ์ เพื่อนำไปใช้วิเคราะห์คำนวณหรือสร้างแผนภูมินำเสนอในลำดับถัดไป

## ความสำคัญของการจัดเตรียมข้อมูลต่อความถูกต้องของผลการวิเคราะห์
- หลักการคุณภาพของข้อมูล (Garbage In, Garbage Out): หากนำตารางข้อมูลต้นฉบับที่มีข้อผิดพลาดมาคำนวณทันที ผลลัพธ์ที่ได้ย่อมคลาดเคลื่อนจากความเป็นจริง แม้ว่าจะเขียนคำสั่งคำนวณได้ถูกต้องตามหลักไวยากรณ์ก็ตาม
- ผลกระทบต่อการคำนวณยอดรวมและค่าเฉลี่ย: การบันทึกข้อมูลซ้ำซ้อนทำให้นับจำนวนคำสั่งซื้อและยอดขายรวมเกินจริง ส่วนค่าผิดปกติเพียงรายการเดียว (เช่น 99,000 บาท) ส่งผลให้ค่าเฉลี่ยคลาดเคลื่อนจาก 754.83 บาท ไปเป็น 3,613.82 บาท
- ผลกระทบต่อการจัดกลุ่มและสรุปรายงาน: อักขระช่องว่างส่วนเกินต่อท้ายข้อความทำให้โปรแกรมแยกนับหมวดหมู่สินค้าเดียวกันออกเป็นคนละกลุ่ม ส่งผลให้ตารางสรุปจำนวนสินค้าแยกตามหมวดหมู่ขาดความถูกต้อง

## ปัญหาคุณภาพข้อมูล 4 รูปแบบหลักที่ต้องตรวจสอบในขั้นตอน Transform
- ข้อมูลสูญหายหรือค่าว่าง (Missing Values): ข้อมูลที่ไม่ได้ถูกบันทึกในบางคอลัมน์ (แสดงเป็น NaN หรือ <NA>) ผู้วิเคราะห์ต้องพิจารณาแยกแยะระหว่างคอลัมน์สำคัญที่ใช้คำนวณยอดขาย กับคอลัมน์ข้อมูลทั่วไปที่สามารถคงแถวข้อมูลไว้ได้
- การบันทึกข้อมูลซ้ำซ้อน (Duplicate Data): รายการคำสั่งซื้อเดียวกันถูกบันทึกซ้ำมากกว่า 1 แถวในทุกคอลัมน์ ซึ่งต้องตรวจสอบแถวที่ซ้ำและตัดออกให้เหลือเพียงรายการจริงเพียงรายการเดียว
- รูปแบบข้อความไม่เป็นมาตรฐานเดียวกัน (Inconsistent Text): ข้อมูลข้อความที่มีอักขระช่องว่างส่วนเกินด้านหน้าหรือด้านหลัง เช่น 'อุปกรณ์ต่อพ่วง ' ซึ่งต้องปรับรูปแบบข้อความให้เป็นมาตรฐานเดียวกันก่อนจัดกลุ่มข้อมูล
- ค่าผิดปกติหรือค่าที่อยู่นอกเกณฑ์ที่กำหนด (Outliers / Invalid Values): ข้อมูลตัวเลขที่ไม่สอดคล้องกับความเป็นจริง เช่น จำนวนชิ้นที่ติดลบ (-2 ชิ้น) หรือยอดเงินที่สูงผิดปกติจากการบันทึกข้อมูลผิดพลาด (99,000 บาท)

## หลักปฏิบัติมาตรฐานในการจัดเตรียมข้อมูลด้วยไลบรารี pandas
- การตรวจสอบข้อมูลก่อนดำเนินการแก้ไข (Inspect Before Transform): ใช้การคัดกรองตามเงื่อนไข (Boolean Indexing) เพื่อแสดงเฉพาะแถวที่มีปัญหาขึ้นมาตรวจสอบรายละเอียดก่อนสั่งลบหรือแก้ไขข้อมูล
- การแยกตัวแปรเพื่อรักษาตารางข้อมูลต้นฉบับ (Preserve Original Data): ควรจัดเก็บตารางข้อมูลที่ผ่านการลบแถวซ้ำซ้อนและคัดกรองค่าว่างไว้ในตัวแปรใหม่ชื่อ df_clean เพื่อเก็บรักษาตารางข้อมูลต้นฉบับ (df) ไว้ใช้ตรวจสอบเปรียบเทียบ
- การเลือกลบค่าว่างเฉพาะคอลัมน์สำคัญ (Selective Cleaning): ระบุพารามิเตอร์ subset ในคำสั่ง .dropna() เพื่อลบเฉพาะแถวที่ขาดข้อมูลจำนวนชิ้นหรือยอดเงินรวม โดยไม่ลบรายการที่ขาดเพียงหมายเลขโทรศัพท์
- การส่งออกไฟล์ข้อมูลโดยไม่บันทึกลำดับแถว (Clean Export): เมื่อบันทึกไฟล์ใหม่ด้วยคำสั่ง .to_csv() ต้องกำหนดพารามิเตอร์ index=False เสมอ เพื่อป้องกันไม่ให้ลำดับแถวถูกบันทึกเป็นคอลัมน์เกินมาในไฟล์ CSV

## ขั้นตอนที่ 9: การคัดกรองขอบเขตข้อมูลด้วยคำสั่ง .between() และการเปรียบเทียบค่าเฉลี่ย

ใช้คำสั่ง .between(1, 20) เพื่อคัดกรองรายการที่มีจำนวนชิ้นตั้งแต่ 1 ถึง 20 ชิ้น และมียอดเงินรวมไม่เกิน 10,000 บาท จากนั้นเปรียบเทียบค่าเฉลี่ยก่อนและหลังการจัดเตรียมข้อมูล

```python
df_clean = df_clean[df_clean['quantity'].between(1, 20) & (df_clean['total_price'] <= 10000)]
```

- `& (df_clean['total_price'] <= 10000)`: ตัวดำเนินการเชื่อมเงื่อนไขว่ายอดเงินรวมต่อรายการต้องไม่เกิน 10,000 บาท
- `round(..., 2)`: ฟังก์ชันปัดเศษทศนิยม 2 ตำแหน่งเพื่อเปรียบเทียบค่าเฉลี่ยของตาราง df และ df_clean

## ขั้นตอนที่ 8: เทคนิคที่ 5: การแก้ไขค่าติดลบด้วยค่าสัมบูรณ์ (.abs) และการจำกัดเพดานค่า (.clip)

สาธิต .abs() เพื่อแปลงค่าติดลบเป็นค่าบวก และ .clip(upper=10) เพื่อจำกัดจำนวนไม่เกิน 10 แก้วตามกติกาสมมติของแบบฝึก จากนั้นแปลงเป็นจำนวนเต็มด้วย .astype(int) การจำกัดเพดานไม่ใช่การกู้คืนค่าจริงจากต้นทาง

```python
branch_df['quantity'] = branch_df['quantity'].abs().clip(upper=10).astype(int)
branch_df['quantity'].describe()
```

- `.abs()`: ฟังก์ชันค่าสัมบูรณ์ (Absolute Value) เปลี่ยนตัวเลขติดลบ (เช่น -2, -3) ให้เป็นตัวเลขบวก (2, 3)
- `.clip(upper=10)`: ฟังก์ชันจำกัดขอบเขตข้อมูล โดยปรับค่าที่สูงเกินเพดาน 10 ให้ลงมาอยู่ที่ค่าสูงสุดที่ยอมรับได้คือ 10
- `.astype(int)`: แปลงชนิดข้อมูลจำนวนชิ้นจากทศนิยม (float64) ให้เป็นจำนวนเต็ม (int64)
- `branch_df['quantity'].describe()`: ตรวจสอบค่าสถิติพื้นฐานของคอลัมน์จำนวนชิ้นภายหลังการปรับแก้

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์

**เงื่อนไขก่อนใช้คำสั่ง:** กติกาสมมติสำหรับฝึกคำสั่ง: ถือว่าค่าติดลบเกิดจากเครื่องหมายผิด และจำกัดจำนวนสูงสุดไว้ที่ 10 แก้วเพื่อสาธิต .clip() การเปลี่ยน 50 เป็น 10 เป็นการจำกัดเพดาน ไม่ใช่การยืนยันยอดขายจริง หากยังตรวจต้นทางไม่ได้ ให้แยกรายการไว้ตรวจสอบ


````python
branch_df['quantity'] = branch_df['quantity'].abs().clip(upper=10).astype(int)
branch_df['quantity'].describe()
````

## ขั้นตอนที่ 9: เทคนิคที่ 6: การลบข้อมูลซ้ำโดยระบุคอลัมน์คีย์หลัก (.drop_duplicates(subset=...))

ลบแถวที่มีรหัสคำสั่งซื้อ (order_id) ซ้ำซ้อนกัน โดยระบุพารามิเตอร์ subset=['order_id'] เพื่อตรวจสอบการซ้ำเฉพาะคอลัมน์คีย์หลัก และเก็บแถวแรกไว้ด้วย keep='first' ลงในตารางใหม่ชื่อ branch_clean

```python
branch_clean = branch_df.drop_duplicates(subset=['order_id'], keep='first').copy()
print('ขนาดตารางก่อนลบซ้ำ:', branch_df.shape)
print('ขนาดตารางหลังลบซ้ำ:', branch_clean.shape)
```

- `branch_df.drop_duplicates(...)`: ฟังก์ชันสำหรับลบแถวข้อมูลที่มีการบันทึกซ้ำซ้อน
- `subset=['order_id']`: ระบุให้ตรวจสอบความซ้ำซ้อนเฉพาะคอลัมน์รหัสคำสั่งซื้อ (order_id) แม้คอลัมน์อื่นอาจบันทึกต่างกันเล็กน้อย
- `keep='first'`: เก็บรักษาแถวแรกของรหัสคำสั่งซื้อนั้นไว้ และลบแถวที่ซ้ำตามมาออก
- `.copy()`: สร้างสำเนาตารางแยกออกมาเป็นตัวแปร branch_clean อย่างอิสระ

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์

**เงื่อนไขก่อนใช้คำสั่ง:** กำหนดให้ 1 order_id แทน 1 รายการขายในชุดฝึกนี้ จึงเก็บแถวแรกเมื่อพบรหัสซ้ำ ข้อมูลจริงที่หนึ่งคำสั่งซื้อมีหลายสินค้า หรือแถวซ้ำมีค่าขัดแย้งกัน ต้องตรวจสอบก่อนลบ


````python
branch_clean = branch_df.drop_duplicates(subset=['order_id'], keep='first').copy()
print('ขนาดตารางก่อนลบซ้ำ:', branch_df.shape)
print('ขนาดตารางหลังลบซ้ำ:', branch_clean.shape)
````
