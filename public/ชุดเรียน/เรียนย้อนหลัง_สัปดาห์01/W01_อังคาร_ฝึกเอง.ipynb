## ขั้นที่ 8: [การเชื่อมโยงสู่ ETL] .describe() ตรวจสอบค่าทางสถิติพื้นฐานเพื่อค้นหาค่าผิดปกติ (Outlier)

เมธอด .describe() คำนวณค่าสถิติเชิงพรรณนา (count, mean, std, min, 25%, 50%, 75%, max) ของคอลัมน์ตัวเลข เพื่อตรวจสอบขอบเขตข้อมูล

## ขั้นที่ 9: [การเชื่อมโยงสู่ ETL] .isna().sum() นับจำนวนค่าว่างแยกตามรายคอลัมน์
