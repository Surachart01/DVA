## วันจันทร์ — วันจันทร์: ทบทวนและสอบกลางภาค

- ทบทวนขอบเขตสัปดาห์ที่ 1–7 เป็นเวลา 60 นาที
- สอบรายบุคคลเป็นเวลา 120 นาที
- กลางภาค 15 คะแนน: ความรู้ 5 คะแนน และปฏิบัติ 10 คะแนน
- การทบทวนใช้ข้อมูลฝึกเดิม ส่วนการสอบใช้เอกสารที่ผู้สอนแจก


## วันจันทร์ — ทบทวนเส้นทางข้อมูลและเหตุผลของแต่ละขั้น

- Get Data: ระบุแหล่งข้อมูล ความหมายคอลัมน์ และคำถามที่จะตอบ
- ETL: ตรวจปัญหา เลือกวิธีจัดการ และตรวจผลหลังแก้ไข
- Modeling: ตรวจคีย์ เชื่อมตาราง และคำนวณตัวชี้วัด
- อธิบายผลด้วยหลักฐาน พร้อมระบุข้อจำกัดของข้อมูล


## วันจันทร์ — เปิดงานฝึกเดิมและตรวจผลทีละคำสั่ง

- เลือก Notebook สัปดาห์ที่ 3, 6 หรือ 7 ตามหัวข้อที่ต้องทบทวน
- อัปโหลดไฟล์ข้อมูลฝึกที่ตรงกับ Notebook
- เขียนหรือทบทวน pandas ทีละคำสั่ง แล้วรันและตรวจผล
- อธิบายว่าผลแต่ละขั้นเชื่อมกับคำถามและขั้นตอนถัดไปอย่างไร


## วันจันทร์ — ตรวจความพร้อมก่อนการสอบรายบุคคล

- รับข้อสอบฉบับผู้เรียนและข้อมูลสอบตามที่ผู้สอนกำหนด
- ตรวจบัญชี Google การเปิด Colab และวิธีส่งหลักฐาน
- อ่านคำชี้แจงและเกณฑ์ก่อนเริ่มนับเวลาสอบ
- ผู้เรียนอธิบายวิธีคิดและตรวจงานของตนเอง


## วันจันทร์ — สอบกลางภาค 120 นาที

- ความรู้ 5 คะแนน และปฏิบัติบน Colab 10 คะแนน
- หลักฐานการสอบ: คำตอบส่วนความรู้และ Notebook รายบุคคล
- ตรวจชื่อผู้เรียนและความครบถ้วนก่อนส่งตามช่องทางที่กำหนด
- คะแนนกลางภาคบันทึกครั้งเดียวในหมวดการสอบ


## วันอังคาร — วันอังคาร: รับผลย้อนกลับและเลือกจุดที่ต้องทบทวน

- รับข้อเสนอแนะรายจุดจากผู้สอนตามผลการประเมินเดิม
- ระบุว่าติดขัดที่การนำเข้า การตรวจข้อมูล การเชื่อมตาราง หรือการแปลผล
- เลือกหัวข้อที่จะฝึกแก้ไขก่อนหนึ่งเรื่อง
- กิจกรรมวันนี้เป็นการซ่อมเสริม ไม่สอบหรือเก็บคะแนนกลางภาคซ้ำ


## วันอังคาร — เตรียมไฟล์ฝึกและ Notebook สัปดาห์เดิม

- ตรวจคุณภาพข้อมูล: ใช้ W03 และ student_register_dirty.csv
- เชื่อมตารางและคำนวณ: ใช้ W06/W07 และ nvc_store_sales.xlsx
- ผู้สอนและผู้เรียนเปิด Notebook ที่เลือกและอัปโหลดไฟล์เดียวกัน
- ตรวจชื่อไฟล์และรันเซลล์เตรียมข้อมูลตามลำดับ


## วันอังคาร — ซ่อมเสริมการอ่านไฟล์และตรวจตาราง

- อ่านไฟล์ด้วย read_csv() หรือ read_excel() ตามรูปแบบข้อมูล
- รัน head() และตรวจ shape ทีละคำสั่ง
- ตรวจชื่อคอลัมน์และชนิดข้อมูลตามขั้นใน Notebook เดิม
- อธิบายความหมายของตารางที่ได้ก่อนเริ่มแก้ไขข้อมูล


## วันอังคาร — ซ่อมเสริมการจัดเตรียม เชื่อม และสรุปข้อมูล

- เลือกขั้นที่ยังติดขัดจากงาน W03, W06 หรือ W07
- ตรวจผลก่อนแก้ไข แล้วใช้ method เดิมทีละรายการ
- ตรวจจำนวนแถว คีย์ หรือผลสรุปหลังรันแต่ละขั้น
- อธิบายเหตุผลของวิธีที่เลือกและผลที่เปลี่ยนไป


## วันอังคาร — สรุปความก้าวหน้าหลังการซ่อมเสริม

- แสดงหนึ่งคำสั่งที่แก้ไขและผลที่ตรวจแล้ว
- อธิบายเหตุผลด้วยคำของตนเอง
- บันทึกเรื่องที่ยังต้องทบทวนและแนวทางฝึกครั้งถัดไป
- บันทึกความก้าวหน้าแยกจากคะแนนกลางภาคเดิม
