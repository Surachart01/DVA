## ขั้นตอนที่ 6: ตรวจสอบคีย์นอกว่าพบในตารางเมนู

## ขั้นตอนที่ 7: แสดงรายการที่ไม่พบรหัสเมนู

## ขั้นตอนที่ 11: เชื่อมตารางพร้อมตรวจสอบความสัมพันธ์

validate กำหนดให้ตาราง menu ต้องมีคีย์ไม่ซ้ำ ส่วน indicator=True เพิ่มคอลัมน์ _merge บอกที่มาของแต่ละแถว

```python
check = sales.merge(menu, on='menu_id', how='left', validate='many_to_one', indicator=True)
```

- `validate='many_to_one'`: กำหนดว่าตารางเมนูต้องมีคีย์ไม่ซ้ำ
- `indicator=True`: เพิ่มคอลัมน์ _merge บอกที่มาของแถว
- `check =`: เก็บผลไว้ตรวจสอบ

พิมพ์คำสั่งและกดปุ่ม Shift + Enter เพื่อรันคำสั่งและตรวจสอบผลลัพธ์

```text
รันคำสั่งสำเร็จ (ไม่มีข้อความแจ้งข้อผิดพลาด)
```

หากตาราง menu มีคีย์ซ้ำ pandas จะแจ้งข้อผิดพลาดก่อนเกิดการนับซ้ำซ้อนเกินจำนวนจริง

**ข้อควรระวังในการเขียนคำสั่ง:** เขียน many_to_one ด้วยขีดล่างและอยู่ในเครื่องหมายคำพูด

**คำถามตรวจสอบความเข้าใจ:** รันได้โดยไม่เกิด MergeError แสดงว่าตาราง menu ไม่มีคีย์ซ้ำ
